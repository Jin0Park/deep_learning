```python
import torch
import torch.distributions as dist
```

```python
## Let's create a Bernoulli distribution

bernoulli = dist.Bernoulli(probs=torch.tensor([0.5]))
print(bernoulli)

## Sample from the distribution
for _ in range(10):
    print(bernoulli.sample())
```

```python
samples = [bernoulli.sample() for _ in range(1000)]
print(f"Mean: {torch.stack(samples).mean()}")
print(f"Variance: {torch.stack(samples).var()}")
```

```python
## lets create a normal distribution
normal = dist.Normal(loc=torch.tensor(0.0), scale=torch.tensor(1.0))
for _ in range(10):
    print(normal.sample())
```

```python
samples = [normal.sample() for _ in range(1000)]
print(f"Mean: {torch.stack(samples).mean()}")
print(f"Variance: {torch.stack(samples).var()}")
```
